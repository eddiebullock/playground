```python
import pandas as pd

# Load processed datasets
ybt = pd.read_csv('/Users/eb2007/playground/bullpy/c4_play2/data/processed/YBT_processed.csv')
c4 = pd.read_csv('/Users/eb2007/playground/bullpy/c4_play2/data/processed/data_c4_processed.csv')
```

```python
# Get means and stds for all numeric columns
ybt_stats = ybt.describe().T[['mean', 'std']]
c4_stats = c4.describe().T[['mean', 'std']]

# Align columns for comparison
common_cols = ybt_stats.index.intersection(c4_stats.index)
ybt_stats = ybt_stats.loc[common_cols]
c4_stats = c4_stats.loc[common_cols]

# Combine for side-by-side comparison
comparison = pd.concat(
    [ybt_stats.add_prefix('YBT_'), c4_stats.add_prefix('C4_')],
    axis=1
)

# Show the first 20 features for a quick look
comparison.head(20)
```

```python
# Autism prevalence breakdown for YBT
print("YBT autism_target breakdown:")
print(ybt['autism_target'].value_counts())
print(ybt['autism_target'].value_counts(normalize=True).mul(100).round(2).astype(str) + '%')

# Autism prevalence breakdown for C4
print("\nC4 autism_target breakdown:")
print(c4['autism_target'].value_counts())
print(c4['autism_target'].value_counts(normalize=True).mul(100).round(2).astype(str) + '%')

# Combined summary
ybt_aut = ybt['autism_target'].mean() * 100
c4_aut = c4['autism_target'].mean() * 100
print(f'\nAutism prevalence: YBT = {ybt_aut:.2f}%, C4 = {c4_aut:.2f}%')
```
